# Step 1: Preview the Adult dataset

**Goal:** display the first few records as a readable table. Each row represents one person, and `income` is the category we will eventually predict.

## 1. Import pandas

```python
import pandas as pd

pd.set_option("display.max_columns", None)
```

## 2. Read the training file and add column names

```python
columns = [
    "age", "workclass", "fnlwgt", "education", "education-num",
    "marital-status", "occupation", "relationship", "race", "sex",
    "capital-gain", "capital-loss", "hours-per-week", "native-country", "income"
]

adult = pd.read_csv(
    "DSDA310/adult.data",
    header=None,
    names=columns,
    skipinitialspace=True
)
```

## 3. Display the first five rows

```python
adult.head()
```

## 4. Preview the separate test file


```python
adult_test = pd.read_csv(
    "DSDA310/adult.test",
    header=None,
    names=columns,
    skipinitialspace=True,
    comment="|"
)

adult_test.head()
```

## 5. Check the table sizes


```python
print("Training file:", adult.shape[0], "rows and", adult.shape[1], "columns")
print("Test file:", adult_test.shape[0], "rows and", adult_test.shape[1], "columns")
```

## What to notice

- `age`, `capital-gain`, and `hours-per-week` contain numbers.
- `workclass`, `education`, and `occupation` contain category labels.
- `income` is the answer we will eventually predict: `<=50K` or `>50K`.
- `education-num` is an ordered education-level code, not a literal count of school years.
