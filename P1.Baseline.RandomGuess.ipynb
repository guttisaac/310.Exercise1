```python
import pandas as pd
columns = [
    "age", "workclass", "fnlwgt", "education", "education_num",
    "marital_status", "occupation", "relationship", "race", "sex",
    "capital_gain", "capital_loss", "hours_per_week", "native_country", "income"
]
df = pd.read_csv(r"C:\Users\gutti\DSDA310\Dataset\adult.data",names=columns, skipinitialspace=True)

y_true = df["income"].map({">50K": 1, "<=50K": 0})
print(df.shape)
print(y_true.value_counts(normalize=True))
```

```python
import numpy as np
from sklearn.metrics import accuracy_score, precision_score, recall_score

p_over = y_true.mean()
rng = np.random.default_rng()
y_pred_random = rng.choice([1, 0], size=len(y_true), p=[p_over, 1 - p_over])

print("Accuracy: ", accuracy_score(y_true, y_pred_random))
print("Precision:", precision_score(y_true, y_pred_random))
print("Recall:   ", recall_score(y_true, y_pred_random))
```
