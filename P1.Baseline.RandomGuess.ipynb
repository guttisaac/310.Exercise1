## 1. Import pandas 


```python
import pandas as pd

pd.set_option("display.max_columns", None)
```

## 2. Load `adult.data` and `adult.test`

```python
columns = [
    "age", "workclass", "fnlwgt", "education", "education_num",
    "marital_status", "occupation", "relationship", "race", "sex",
    "capital_gain", "capital_loss", "hours_per_week", "native_country", "income",
]

def load_adult(filename, is_test_file=False):
    data = pd.read_csv(
        filename,
        header=None,
        names=columns,
        skipinitialspace=True,
        na_values="?",
        comment="|" if is_test_file else None,
    )

    # Remove spaces around text. The test file also has a final period in its labels.
    for column in data.select_dtypes(include="object").columns:
        data[column] = data[column].str.strip()
    data["income"] = data["income"].str.rstrip(".")
    return data

development_data = load_adult("DSDA310/adult.data")
test_data = load_adult("DSDA310/adult.test", is_test_file=True)

print("Development file shape:", development_data.shape)
print("Test file shape:       ", test_data.shape)
print("Development labels:    ", sorted(development_data["income"].unique()))
print("Test labels:           ", sorted(test_data["income"].unique()))
```
